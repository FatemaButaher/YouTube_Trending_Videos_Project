```python
def category_name(ID): 
    if ID == 24 : 
        return "Entertainment"
    elif ID == 10 :
        return "Music"
    elif ID == 22 :
        return "People & Blogs"
    elif ID == 23 :
        return "Comedy"
    elif ID == 25 :
        return "News & Politics"
    elif ID == 17 :
        return "Sports"
    elif ID == 26 :
        return "Howto & Style"
    elif ID == 1 :
        return "Film & Animation"
    elif ID == 20 :
        return "Gaming"
    elif ID == 28 :
        return "Science & Technology"
    elif ID == 27 :
        return "Education"
    elif ID == 15 :
        return "Pets & Animals"
    elif ID == 2 :
        return "Autos & Vehicles"
    elif ID == 19 :
        return "Travel & Events"
    elif ID == 29 :
        return "Nonprofits & Activism"
    elif ID == 43 :
        return "Shows"
    elif ID == 30 :
        return "Movies"
    elif ID == 44 :
        return "Trailers"
```

```python
Youtube_df["category_name"] = Youtube_df["category_id"].apply(category_name)
```

```python
Youtube_df["trending_date"] = pd.to_datetime(Youtube_df["trending_date"], format='%y.%d.%m')
Youtube_df["publish_date"] = pd.to_datetime(Youtube_df["publish_date"])
```

##### ----------------------------------------------------------
##### columns describtion: 
- `title`: The title of the video.
- `channel_title`: The title of the YouTube channel that published the video.
- `publish_date`: The date when the video was published on YouTube.
- `time_frame`: The duration of time (e.g., 1 day, 6 hours) that the video has been trending on YouTube.
- `published_day_of_week`: The day of week (e.g., Monday) when the video was published.
- `publish_country`: The country where the video was published.
- `tags`: The tags or keywords associated with the video.
- `views`: The number of views received by a particular video
- `likes`: Number o likes received per each videos
- `dislike`: Number dislikes receives per an individual vidoe
- `comment_count`: number of comments
- `video_id`: The unique identifier of the video on YouTube
- `trending_date`: The date on which the video appeared on YouTube's trending list.
- `category_id`: The numeric ID of the video's category according to YouTube's category system
- `comments_disabled`:A boolean (True/False) indicating whether comments are disabled for the video.
- `ratings_disabled`: A boolean (True/False) indicating whether ratings (likes and dislikes) are disabled for the video.
- `video_error_or_removed`: A boolean (True/False) indicating whether the video has an error or has been removed.
- `category_name`: The name of the video's category corresponding to category_id.

```python
import matplotlib.pyplot as plt
import numpy as np

selected_categories = ["Entertainment", "Music", "People & Blogs"]
plot_df = Youtube_df[Youtube_df['category_name'].isin(selected_categories)].copy()

plot_df = plot_df.sample(10000, random_state=1)

size = np.sqrt(plot_df['comment_count'])
size = 10 + (size / size.max()) * 600

color_map = {
    "Entertainment": "#e74c3c",   
    "Music": "#2c3e50",          
    "People & Blogs": "#27ae60"  
}

fig, ax = plt.subplots(figsize=(12, 10))

for cat in selected_categories:
    mask = plot_df['category_name'] == cat
    ax.scatter(
        plot_df.loc[mask, 'views'],
        plot_df.loc[mask, 'likes'],
        s=size[mask],
        color=color_map[cat],
        alpha=0.15,
        edgecolors='none',
        label=cat
    )

ax.set_xscale('log')
ax.set_yscale('log')
ax.set_xlabel("Views")
ax.set_ylabel("Likes")
ax.set_title("YouTube Trending: Entertainment, Music, People & Blogs (July-2017 to Aug-2018)")

ax.legend(loc='upper left', title="Category", markerscale=0.5)

plt.tight_layout()
plt.show()
```

```python
from wordcloud import WordCloud
import re

text = " ".join(Youtube_df['tags'].drop_duplicates().astype(str))

english_words = re.findall(r'\b[a-zA-Z]{3,}\b', text)
text_english_only = " ".join(english_words)

wordcloud = WordCloud(width=1000, height=500, background_color="black").generate(text_english_only)

plt.figure(figsize=(12, 6))
plt.imshow(wordcloud)
plt.title("Most Common Words in Video Tags")
plt.axis('off')


plt.show()
```
